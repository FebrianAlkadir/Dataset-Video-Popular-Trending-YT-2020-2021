# Dataset 2 - Analisis Jenis Klasifikasi

Notebook ini menganalisis satu dataset biner, satu multikelas, dan satu multilabel dengan pandas. Semua dataset memiliki minimal 150 baris. Klasifikasi biner memilih tepat satu dari dua kelas; multikelas memilih tepat satu dari lebih dari dua kelas; multilabel dapat mengaktifkan beberapa label sekaligus pada satu baris.

Ringkasan materi: [Notion - Jenis-Jenis Klasifikasi](https://app.notion.com/p/Jenis-Jenis-Klasifikasi-1-405e27a5dc7b80708a7aff22968f64cf).

> Jalankan sel kode secara berurutan. Setiap sel menampilkan ukuran data dan distribusi label dari CSV di folder yang sama.

## 1. Klasifikasi Biner - Banknote Authentication

Empat fitur berupa varians, skewness, kurtosis, dan entropy dari citra uang kertas digunakan untuk mengklasifikasikan keaslian. Kolom Class pada berkas CSV sumber berisi dua kategori (1 dan 2); setiap uang kertas mendapat satu kategori.

- **Target/label:** `Class`
- **Sumber dataset:** [https://archive.ics.uci.edu/dataset/267/banknote+authentication](https://archive.ics.uci.edu/dataset/267/banknote+authentication)
- **Berkas data langsung/referensi:** [CSV sumber](https://raw.githubusercontent.com/datasets/openml-datasets/ac945980fbb83bf4200a893593a6c1099679d150/data/banknote-authentication/banknote-authentication.csv)

```python
import pandas as pd

df = pd.read_csv('banknote_authentication.csv')
target = 'Class'
label_counts = df[target].value_counts(dropna=False).sort_index()

print('Dataset: Banknote Authentication')
print(f"Jumlah baris: {len(df)}")
print(f"Jumlah fitur: {df.shape[1] - 1}")
print(f"Jumlah kategori target: {df[target].nunique(dropna=False)}")
print("Distribusi label:")
print(label_counts.to_string())
assert len(df) >= 150
assert df[target].nunique(dropna=False) == 2
```

## 2. Klasifikasi Multikelas - Iris

Empat ukuran sepal dan petal dipakai untuk mengidentifikasi spesies Iris-setosa, Iris-versicolor, atau Iris-virginica. Satu bunga hanya memiliki satu spesies, dengan tiga kelas yang mungkin, sehingga ini multikelas.

- **Target/label:** `class`
- **Sumber dataset:** [https://archive.ics.uci.edu/dataset/53/iris](https://archive.ics.uci.edu/dataset/53/iris)
- **Berkas data langsung/referensi:** [CSV sumber](https://raw.githubusercontent.com/datasets/openml-datasets/ac945980fbb83bf4200a893593a6c1099679d150/data/iris/iris.csv)

```python
import pandas as pd

df = pd.read_csv('iris.csv')
target = 'class'
label_counts = df[target].value_counts(dropna=False).sort_index()

print('Dataset: Iris')
print(f"Jumlah baris: {len(df)}")
print(f"Jumlah fitur: {df.shape[1] - 1}")
print(f"Jumlah kategori target: {df[target].nunique(dropna=False)}")
print("Distribusi label:")
print(label_counts.to_string())
assert len(df) >= 150
assert df[target].nunique(dropna=False) > 2
```

## 3. Klasifikasi Multilabel - Flags

Data memuat ciri geografi, populasi, bahasa, dan agama dari 194 negara. Tujuh target menandai warna yang muncul pada bendera: red (merah), green (hijau), blue (biru), yellow (kuning), white (putih), black (hitam), dan orange (oranye). Satu bendera dapat memuat beberapa warna, sehingga beberapa label dapat aktif bersamaan.

- **Target/label:** `red`, `green`, `blue`, `yellow`, `white`, `black`, `orange`
- **Sumber dataset:** [https://archive.ics.uci.edu/dataset/40/flags](https://archive.ics.uci.edu/dataset/40/flags)
- **Berkas data langsung/referensi:** [berkas Flags pada Mulan](https://github.com/tsoumakas/mulan/blob/43f6389095ad71984264db4387a8074b22bc7d4a/data/multi-label/flags/flags.arff)

```python
import pandas as pd

df = pd.read_csv('flags.csv')
target_cols = ['red', 'green', 'blue', 'yellow', 'white', 'black', 'orange']
label_matrix = df[target_cols].apply(pd.to_numeric, errors="raise").astype(int)

print('Dataset: Flags')
print(f"Jumlah baris: {len(df)}")
print(f"Jumlah fitur: {df.shape[1] - len(target_cols)}")
print(f"Jumlah label keluaran: {len(target_cols)}")
print("Jumlah sampel per label aktif:")
print(label_matrix.sum().to_string())
print("\nDistribusi jumlah label aktif per sampel:")
print(label_matrix.sum(axis=1).value_counts().sort_index().to_string())
assert len(df) >= 150
assert len(target_cols) >= 2
assert label_matrix.shape[1] == len(target_cols)
```
