# Dataset 3 - Analisis Jenis Klasifikasi

Notebook ini menganalisis satu dataset biner, satu multikelas, dan satu multilabel dengan pandas. Semua dataset memiliki minimal 150 baris. Klasifikasi biner memilih tepat satu dari dua kelas; multikelas memilih tepat satu dari lebih dari dua kelas; multilabel dapat mengaktifkan beberapa label sekaligus pada satu baris.

Ringkasan materi: [Notion - Jenis-Jenis Klasifikasi](https://app.notion.com/p/Jenis-Jenis-Klasifikasi-1-405e27a5dc7b80708a7aff22968f64cf).

> Jalankan sel kode secara berurutan. Setiap sel menampilkan ukuran data dan distribusi label dari CSV di folder yang sama.

## 1. Klasifikasi Biner - Pima Indians Diabetes

Delapan fitur klinis (termasuk kadar glukosa, tekanan darah, BMI, usia, dan jumlah kehamilan) digunakan untuk memprediksi hasil tes diabetes positif atau negatif. Target memiliki tepat dua kategori.

- **Target/label:** `class`
- **Sumber dataset:** [https://archive.ics.uci.edu/dataset/34/diabetes](https://archive.ics.uci.edu/dataset/34/diabetes)
- **Berkas data langsung/referensi:** [CSV sumber](https://raw.githubusercontent.com/datasets/openml-datasets/ac945980fbb83bf4200a893593a6c1099679d150/data/diabetes/diabetes.csv)

```python
import pandas as pd

df = pd.read_csv('pima_diabetes.csv')
target = 'class'
label_counts = df[target].value_counts(dropna=False).sort_index()

print('Dataset: Pima Indians Diabetes')
print(f"Jumlah baris: {len(df)}")
print(f"Jumlah fitur: {df.shape[1] - 1}")
print(f"Jumlah kategori target: {df[target].nunique(dropna=False)}")
print("Distribusi label:")
print(label_counts.to_string())
assert len(df) >= 150
assert df[target].nunique(dropna=False) == 2
```

## 2. Klasifikasi Multikelas - Glass Identification

Sembilan fitur komposisi kimia dan indeks bias digunakan untuk mengidentifikasi jenis kaca. Kolom Type memiliki lebih dari dua jenis kaca, dan satu sampel hanya masuk ke satu jenis, maka ini klasifikasi multikelas.

- **Target/label:** `Type`
- **Sumber dataset:** [https://archive.ics.uci.edu/dataset/42/glass+identification](https://archive.ics.uci.edu/dataset/42/glass+identification)
- **Berkas data langsung/referensi:** [CSV sumber](https://raw.githubusercontent.com/datasets/openml-datasets/ac945980fbb83bf4200a893593a6c1099679d150/data/glass/glass.csv)

```python
import pandas as pd

df = pd.read_csv('glass.csv')
target = 'Type'
label_counts = df[target].value_counts(dropna=False).sort_index()

print('Dataset: Glass Identification')
print(f"Jumlah baris: {len(df)}")
print(f"Jumlah fitur: {df.shape[1] - 1}")
print(f"Jumlah kategori target: {df[target].nunique(dropna=False)}")
print("Distribusi label:")
print(label_counts.to_string())
assert len(df) >= 150
assert df[target].nunique(dropna=False) > 2
```

## 3. Klasifikasi Multilabel - Yeast Protein Localization

Fitur biologis protein yeast dipetakan ke 14 label lokasi/substrat sel (Class1 hingga Class14 pada berkas sumber). Satu protein dapat memiliki beberapa lokasi sekaligus; karena itu keluaran dapat berisi banyak label aktif, bukan satu kelas tunggal.

- **Target/label:** `Class1`, `Class2`, `Class3`, `Class4`, `Class5`, `Class6`, `Class7`, `Class8`, `Class9`, `Class10`, `Class11`, `Class12`, `Class13`, `Class14`
- **Sumber dataset:** [https://github.com/tsoumakas/mulan/tree/43f6389095ad71984264db4387a8074b22bc7d4a/data/multi-label/yeast](https://github.com/tsoumakas/mulan/tree/43f6389095ad71984264db4387a8074b22bc7d4a/data/multi-label/yeast)
- **Berkas data langsung/referensi:** [berkas Yeast pada Mulan](https://github.com/tsoumakas/mulan/blob/43f6389095ad71984264db4387a8074b22bc7d4a/data/multi-label/yeast/yeast.arff)

```python
import pandas as pd

df = pd.read_csv('yeast.csv')
target_cols = ['Class1', 'Class2', 'Class3', 'Class4', 'Class5', 'Class6', 'Class7', 'Class8', 'Class9', 'Class10', 'Class11', 'Class12', 'Class13', 'Class14']
label_matrix = df[target_cols].apply(pd.to_numeric, errors="raise").astype(int)

print('Dataset: Yeast Protein Localization')
print(f"Jumlah baris: {len(df)}")
print(f"Jumlah fitur: {df.shape[1] - len(target_cols)}")
print(f"Jumlah label keluaran: {len(target_cols)}")
print("Jumlah sampel per label aktif:")
print(label_matrix.sum().to_string())
print("\nDistribusi jumlah label aktif per sampel:")
print(label_matrix.sum(axis=1).value_counts().sort_index().to_string())
assert len(df) >= 150
assert len(target_cols) >= 2
assert label_matrix.shape[1] == len(target_cols)
```
