# Dataset 1 - Analisis Jenis Klasifikasi

Notebook ini menganalisis satu dataset biner, satu multikelas, dan satu multilabel dengan pandas. Semua dataset memiliki minimal 150 baris. Klasifikasi biner memilih tepat satu dari dua kelas; multikelas memilih tepat satu dari lebih dari dua kelas; multilabel dapat mengaktifkan beberapa label sekaligus pada satu baris.

Ringkasan materi: [Notion - Jenis-Jenis Klasifikasi](https://app.notion.com/p/Jenis-Jenis-Klasifikasi-1-405e27a5dc7b80708a7aff22968f64cf).

> Jalankan sel kode secara berurutan. Setiap sel menampilkan ukuran data dan distribusi label dari CSV di folder yang sama.

## 1. Klasifikasi Biner - Wisconsin Diagnostic Breast Cancer

Setiap baris mewakili sampel tumor payudara dengan 30 fitur pengukuran inti sel. Target diagnosis hanya memiliki dua nilai: B (benign/jinak) dan M (malignant/ganas). Model memilih tepat satu dari dua kelas tersebut, jadi ini klasifikasi biner.

- **Target/label:** `diagnosis`
- **Sumber dataset:** [https://archive.ics.uci.edu/dataset/17/breast+cancer+wisconsin+diagnostic](https://archive.ics.uci.edu/dataset/17/breast+cancer+wisconsin+diagnostic)
- **Berkas data langsung/referensi:** [data asli](https://archive.ics.uci.edu/ml/machine-learning-databases/breast-cancer-wisconsin/wdbc.data)

```python
import pandas as pd

df = pd.read_csv('breast_cancer.csv')
target = 'diagnosis'
label_counts = df[target].value_counts(dropna=False).sort_index()

print('Dataset: Wisconsin Diagnostic Breast Cancer')
print(f"Jumlah baris: {len(df)}")
print(f"Jumlah fitur: {df.shape[1] - 1}")
print(f"Jumlah kategori target: {df[target].nunique(dropna=False)}")
print("Distribusi label:")
print(label_counts.to_string())
assert len(df) >= 150
assert df[target].nunique(dropna=False) == 2
```

## 2. Klasifikasi Multikelas - Wine Recognition

Dataset berisi 13 pengukuran kimia dari sampel wine. Label kelas 1, 2, dan 3 mewakili tiga varietas/kultivar wine. Setiap sampel hanya memiliki satu label, tetapi pilihannya lebih dari dua, sehingga ini multikelas.

- **Target/label:** `class`
- **Sumber dataset:** [https://archive.ics.uci.edu/dataset/109/wine](https://archive.ics.uci.edu/dataset/109/wine)
- **Berkas data langsung/referensi:** [data asli](https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data)

```python
import pandas as pd

df = pd.read_csv('wine.csv')
target = 'class'
label_counts = df[target].value_counts(dropna=False).sort_index()

print('Dataset: Wine Recognition')
print(f"Jumlah baris: {len(df)}")
print(f"Jumlah fitur: {df.shape[1] - 1}")
print(f"Jumlah kategori target: {df[target].nunique(dropna=False)}")
print("Distribusi label:")
print(label_counts.to_string())
assert len(df) >= 150
assert df[target].nunique(dropna=False) > 2
```

## 3. Klasifikasi Multilabel - Emotions

Fitur akustik lagu (72 fitur) digunakan untuk mengenali enam label emosi. Labelnya adalah amazed-suprised (kagum/terkejut), happy-pleased (senang/puas), relaxing-calm (relaks/tenang), quiet-still (hening), sad-lonely (sedih/kesepian), dan angry-aggresive (marah/agresif). Satu lagu dapat memiliki beberapa emosi sekaligus. CSV ini menggabungkan partisi train1 dan test1.

- **Target/label:** `amazed-suprised`, `happy-pleased`, `relaxing-calm`, `quiet-still`, `sad-lonely`, `angry-aggresive`
- **Sumber dataset:** [https://github.com/kdis-lab/ExecuteMulan/tree/f0c9bf5de82b60cc670c8f07eb16927eda74b66e/data/emotions](https://github.com/kdis-lab/ExecuteMulan/tree/f0c9bf5de82b60cc670c8f07eb16927eda74b66e/data/emotions)
- **Berkas data langsung/referensi:** [train1](https://raw.githubusercontent.com/kdis-lab/ExecuteMulan/f0c9bf5de82b60cc670c8f07eb16927eda74b66e/data/emotions/emotions-train1.arff) dan [test1](https://raw.githubusercontent.com/kdis-lab/ExecuteMulan/f0c9bf5de82b60cc670c8f07eb16927eda74b66e/data/emotions/emotions-test1.arff)

```python
import pandas as pd

df = pd.read_csv('emotions.csv')
target_cols = ['amazed-suprised', 'happy-pleased', 'relaxing-calm', 'quiet-still', 'sad-lonely', 'angry-aggresive']
label_matrix = df[target_cols].apply(pd.to_numeric, errors="raise").astype(int)

print('Dataset: Emotions')
print(f"Jumlah baris: {len(df)}")
print(f"Jumlah fitur: {df.shape[1] - len(target_cols)}")
print(f"Jumlah label keluaran: {len(target_cols)}")
print("Jumlah sampel per label aktif:")
print(label_matrix.sum().to_string())
print("\nDistribusi jumlah label aktif per sampel:")
print(label_matrix.sum(axis=1).value_counts().sort_index().to_string())
assert len(df) >= 150
assert len(target_cols) >= 2
assert label_matrix.shape[1] == len(target_cols)
```
